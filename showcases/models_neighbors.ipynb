# K-Nearest Neighbors

```python
import sys
from pathlib import Path

ROOT = Path.cwd()
if ROOT.name == "showcases":
    ROOT = ROOT.parent

sys.path.insert(0, str(ROOT))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ml_code.synthetic_data import make_regression, make_classification
from ml_code.model_selection import random_split
from ml_code.metrics import r_squared, accuracy
```

```python
# Prepare data for regression and classification
x, y = make_classification(n_samples=500, n_features=5, n_classes=2, seed=42)
x_tr_cl, x_te_cl, y_tr_cl, y_te_cl = random_split(x, y, seed=42)

x, y = make_regression(n_samples=500, n_features=5, seed=42)
x_tr_reg, x_te_reg, y_tr_reg, y_te_reg = random_split(x, y, seed=42)
```

## KNNClassifier

```python
from ml_code.models import KNNClassifier

knn_clf = KNNClassifier()
print(knn_clf)

knn_clf.train(x_tr_cl, y_tr_cl)
knn_clf_pred = knn_clf(x_te_cl)
print("Accuracy:", accuracy(y_te_cl, knn_clf_pred))
```

## KNNRegressor

```python
from ml_code.models import KNNRegressor

knn_reg = KNNRegressor()
print(knn_reg)

knn_reg.train(x_tr_reg, y_tr_reg)
knn_reg_pred = knn_reg(x_te_reg)
print("R-squared:", r_squared(y_te_reg, knn_reg_pred))
```
