```python
import numpy as np
import matplotlib.pyplot as plt
```

## Linear data for regression

```python
from ml_code.synthetic_data import make_regression

plt.figure(figsize=(10, 10))

x, y = make_regression(function='linear', n_samples=200, n_features=1, noise_coef=0.1, seed=42)
plt.subplot(2, 2, 1)
plt.scatter(x, y)
plt.title('Linear dataset example')

x, y = make_regression(function='sin', n_samples=200, n_features=1, noise_coef=0.1, seed=42)
plt.subplot(2, 2, 2)
plt.scatter(x, y)
plt.title('Sin dataset example')

x, y = make_regression(function='polynomial', n_samples=200, n_features=1, noise_coef=0.1, seed=42)
plt.subplot(2, 2, 3)
plt.scatter(x, y)
plt.title('Polynomial dataset example')

x, y = make_regression(function='mixed', n_samples=200, n_features=1, noise_coef=0.1, seed=42)
plt.subplot(2, 2, 4)
plt.scatter(x, y)
plt.title('Mixed dataset example');
```

## Data for binary/multiclass classification
