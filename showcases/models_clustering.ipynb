## K-Means

```python
from ml_code.models import KMeans
from sklearn.datasets import make_blobs

x, y = make_blobs(n_samples=100, n_features=2, centers=3, random_state=42)

kmeans = KMeans(n_clusters=3, n_inits=10, max_iter=10)
print(kmeans)
kmeans.train(x)
predicted_labels = kmeans(x)

plt.figure(figsize=(8, 3))
plt.subplot(1, 2, 1)
plt.scatter(x[:, 0], x[:, 1], c=y)
plt.title('True labels')

plt.subplot(1, 2, 2)
plt.scatter(x[:, 0], x[:, 1], c=predicted_labels)
plt.title('K-Means labels');
```
